```python
import random
names_list = ["JOHN", "PAUL", "GEORGE", "RINGO"]
```

```python
class Beetle:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp
        self.max_damage = 10

    def info(self):
        return "Жук " + self.name + " (HP: " + str(self.hp) + ")"

    def get_style(self):
        if self.name == "JOHN":
            return "в стиле Джонни"
        elif self.name == "PAUL":
            return "в стиле Маккартни"
        elif self.name == "GEORGE":
            return "в стиле Харрисона"
        elif self.name == "RINGO":
            return "в стиле Ринго"
        else:
            return "просто так"

    def do_attack(self, enemy):
        damage = random.randint(1, self.max_damage)
        print(self.name + " бьет " + enemy.name + " " + self.get_style() + " на " + str(damage))
        enemy.hp = enemy.hp - damage
```

```python
class BeetlesArmy:
    def __init__(self, army_name, count, max_hp):
        self.army_name = army_name
        self.zhuki = []

        for i in range(count):
            new_hp = random.randint(10, max_hp)
            odin_zhuk = Beetle(army_name, new_hp)
            self.zhuki.append(odin_zhuk)

    def is_army_alive(self):
        if len(self.zhuki) > 0:
            return True
        else:
            return False
```

```python
def start_battle(army1, army2):
    print("--- БИТВА НАЧИНАЕТСЯ: " + army1.army_name + " VS " + army2.army_name + " ---")

    round_num = 1
    while army1.is_army_alive() and army2.is_army_alive():
        print("РАУНД " + str(round_num))

        fighter1 = army1.zhuki[0]
        fighter2 = army2.zhuki[0]

        fighter1.do_attack(fighter2)
        if fighter2.hp <= 0:
            print("Жук " + fighter2.name + " погиб!")
            army2.zhuki.pop(0)

        if army2.is_army_alive():
            fighter2.do_attack(fighter1)
            if fighter1.hp <= 0:
                print("Жук " + fighter1.name + " погиб!")
                army1.zhuki.pop(0)

        print("Осталось жуков: " + army1.army_name + ": " + str(len(army1.zhuki)) + ", " + army2.army_name + ": " + str(len(army2.zhuki)))
        print("--------------------")
        round_num = round_num + 1

    if army1.is_army_alive():
        print("ПОБЕДИЛА АРМИЯ: " + army1.army_name)
    else:
        print("ПОБЕДИЛА АРМИЯ: " + army2.army_name)
```

```python
name1 = input("Первая армия: ").upper()
size1 = int(input("Кол-во жуков: "))
hp1 = int(input("Макс. HP: "))

name2 = input("Вторая армия: ").upper()
size2 = int(input("Кол-во жуков: "))
hp2 = int(input("Макс. HP: "))

army_one = BeetlesArmy(name1, size1, hp1)
army_two = BeetlesArmy(name2, size2, hp2)

start_battle(army_one, army_two)
```
