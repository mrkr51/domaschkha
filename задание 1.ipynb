```python
import requests
from rich.console import Console
from rich.panel import Panel
from rich.markdown import Markdown

console = Console()

API_KEY = "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"
URL = "https://openrouter.ai/api/v1/chat/completions"

messages = [
    {
        "role": "system",
        "content": (
            "Ты это Наталья Морская Пехота как в одноименном меме. Ты наглая бабка в автобусе, которая ругается с водителем. Наглая дерзкая, можешь оскорбить, у тебя нет денег на проезд и есть джип в москве.  "
        ),
    },
]

def get_response():
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "HTTP-Referer": "http://localhost:3000",
        "X-Title": "Natalya Marine Chat",
        "Content-Type": "application/json"
    }

    data = {
        "model": "meta-llama/llama-3.3-70b-instruct:free",
        "messages": messages,
        "temperature": 0.9,
        "max_tokens": 1000
    }

    try:
        response = requests.post(URL, headers=headers, json=data, timeout=30)

        if response.status_code == 200:
            assistant_message = response.json()['choices'][0]['message']['content']
            messages.append({"role": "assistant", "content": assistant_message})
            return assistant_message
        else:
            return f"ОШИБКА: нет денег на проезд >:( {response.status_code}"

    except Exception as e:
        return f"ОШИБКА: нет денег на проезд >:( {e}"

def start_marine_chat():
    console.print(Panel(
        "[bold white]Наталья могрская пехота!Я сказала стартуем![/bold white]\n\n"
        "Команды:\n"
        "• [cyan]/exit[/cyan] - десантироваться (выход)\n"
        "• [cyan]/clear[/cyan] - забыть всё и поехать заново",
        title="[bold blue]Питерский автобус[/bold blue]",
        border_style="bold blue"
    ))

    while True:
        user_input = console.input("\n[bold yellow]Оператор:[/bold yellow] ").strip()

        if user_input.lower() == "/exit":
            console.print("[bold blue]Вы вылетели из автобуса потому что Наталья все таки встала... [/bold blue]")
            break

        if user_input.lower() == "/clear":
            global messages
            messages = [messages[0]]
            console.print("[blue]уезжаем в Москву, у меня там джип.[/blue]")
            continue

        if not user_input:
            continue

        messages.append({"role": "user", "content": user_input})

        with console.status("[bold blue]Наталья просит мелочь на проезд...[/bold blue]"):
            response = get_response()

        if response:
            console.print(Panel(
                Markdown(response),
                title="[bold red]Наталья высказывается:[/bold red]",
                border_style="blue"
            ))

if __name__ == "__main__":
    start_marine_chat()
```
