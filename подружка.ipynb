```python
from agno.agent import Agent
from agno.models.openai import OpenAIChat

my_api_key = "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"
my_base_url = "https://openrouter.ai/api/v1"
my_model_id = "meta-llama/llama-3.3-70b-instruct:free"

model = OpenAIChat(
    id=my_model_id,
    base_url=my_base_url,
    api_key=my_api_key,
)

def toxic_bestie(user_input):
    stop_words = ["учеба", "экзамен", "отчет", "курсовая", "работа"]
    for word in stop_words:
        if word in user_input.lower():
            return "Зай, ну ты чего такая душная? Давай не будем о скучном, у меня от этого голова болит."

    prompt = f"""
    Ты моя токсичная "лучшая" подружка.
    Твоя цель говорить только о себе и о своем богатом муже которого ты называешь Масиком, тебе больше ничего не интересно. ты очень много говоришь о себе или о масике, прям так, чтобы можно было м инут 10 слушать

    Правила:
    1. На мои слова реагируй коротко: "ого", "ясно".
    2. Рассказывай только о своей богатой жизни и о том что для тебя делает твой муж.
    3. обращайся ко мне как "зай" или "дорогуша".

    Примеры:
    Я: Купила платье.
    Ты: Прикольно. А Масик мне сказал, что я достойна только кутюр!

    Мое сообщение: {user_input}
    Твой ответ:
    """

    agent = Agent(model=model)
    response = agent.run(prompt)
    return response.content.strip()

while True:
    user_text = input("\nТы пишешь: ")
    if user_text.lower() in ["стоп", "пока"]:
        break
    print(toxic_bestie(user_text))
```
